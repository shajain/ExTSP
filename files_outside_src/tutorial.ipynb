### Extract extsp scores derived from GTEx

```python
#To get the datasets for a disease run the following
# disases = ["CM", "PCD", "PH", "PCD", "ASD", "nonTarget"]
from ExTSP.tripletSets_exTSP import get_tripletSets_with_exTSP, variantCountsWithDuplicates
# All CM variante
df_target_all = get_tripletSets_with_exTSP("CM")
# CM P variant
df_target_p = get_tripletSets_with_exTSP("CM", "P")
print(f"CM P variants: {variantCountsWithDuplicates(df_target_p)}")
# CM B variant
df_target_b = get_tripletSets_with_exTSP("CM", "B")
print(f"CM B variants: {variantCountsWithDuplicates(df_target_b)}")
# CM PLP variant    
df_target_plp = get_tripletSets_with_exTSP("CM", "PLP")
print(f"CM PLP variants: {variantCountsWithDuplicates(df_target_plp)}")
# CM BLB variant
df_target_blb = get_tripletSets_with_exTSP("CM", "BLB")
print(f"CM BLB variants: {variantCountsWithDuplicates(df_target_blb)}")
# nonTarget PLP variant
df_nonTarget_plp = get_tripletSets_with_exTSP("nonTarget", "PLP")
print(f"nonTarget PLP variants: {variantCountsWithDuplicates(df_nonTarget_plp)}")
# nonTarget BLB variant
df_nonTarget_blb = get_tripletSets_with_exTSP("nonTarget", "BLB")
print(f"nonTarget BLB variants: {variantCountsWithDuplicates(df_nonTarget_blb)}")
# ASD cases gtex
df_ASD_cases = get_tripletSets_with_exTSP("ASD", "case")
print(f"ASD cases variants: {variantCountsWithDuplicates(df_ASD_cases)}")
#ASD controls gtex
df_ASD_controls = get_tripletSets_with_exTSP("ASD", "control")
print(f"ASD controls variants: {variantCountsWithDuplicates(df_ASD_controls)}")
#ASD cases brainspan
df_ASD_cases_brainspan = get_tripletSets_with_exTSP("ASD", "case", brainspan=True)
print(f"ASD cases brainspan variants: {variantCountsWithDuplicates(df_ASD_cases_brainspan)}")
#ASD controls brainspan
df_ASD_controls_brainspan = get_tripletSets_with_exTSP("ASD", "control", brainspan=True)
print(f"ASD controls brainspan variants: {variantCountsWithDuplicates(df_ASD_controls_brainspan)}")










```

#### To get same number of of benign variants as pathogenic by duplication use **numVars** argument as follows

```python
from ExTSP.tripletSets_exTSP import get_tripletSets_with_exTSP, variantCountsWithDuplicates
# CM PLP variant    
df_target_plp = get_tripletSets_with_exTSP("CM", "PLP")
print(f"CM PLP variants: {variantCountsWithDuplicates(df_target_plp)}")
# CM BLB variant
df_target_blb = get_tripletSets_with_exTSP("CM", "BLB", numVars=df_target_plp.Variant.nunique())
print(f"CM BLB variants: {variantCountsWithDuplicates(df_target_blb)}")
```
